```python
# python question : while loop with continue and break 
# write a python program that repeatedly asks the user to enter an integer
#if the user enters a negative number, use continue to ingnore it and ask for another number
#if the user enters 0 , use break to stop the loop
# for every positive number, print its square
# after the loop ends , print "program ended" 
while True:
    num = int(input("Enter an integer: "))

    if num < 0:
        continue

    if num == 0:
        break

    print("Square:", num ** 2)

print("program ended")
```
