```python
product_first = 4 * 3
default_result = 7 + product_first       # 7 + (4 * 3)
sum_first = 7 + 4
parenthesized_result = sum_first * 3      # (7 + 4) * 3
print('7 + 4 * 3: multiply first -> 7 + 12 ->', default_result)
print('(7 + 4) * 3: parentheses first -> 11 * 3 ->', parenthesized_result)

inner_power = 3 ** 2
right_associative = 2 ** inner_power      # 2 ** (3 ** 2)
forced_left_power = (2 ** 3) ** 2
print('2 ** 3 ** 2: compute 3 ** 2 first -> 2 ** 9 ->', right_associative)
print('(2 ** 3) ** 2: parentheses force 2 ** 3 first -> 8 ** 2 ->', forced_left_power)
```

```python
first_division = 6 / 3
left_associative_result = first_division / 2
print('6 / 3 / 2 -> (6 / 3) / 2 -> 2.0 / 2 ->', left_associative_result)
print('6 / (3 / 2) would instead be ->', 6 / (3 / 2))

# A tie between powers: ** groups from right to left.
right_inner = 3 ** 2
power_result = 2 ** right_inner
print('2 ** 3 ** 2 -> 2 ** (3 ** 2) -> 2 ** 9 ->', power_result)
print('(2 ** 3) ** 2 forces the other grouping ->', (2 ** 3) ** 2)

# Mixed arithmetic: *, /, //, and % share a level; + and - come later.
parenthesized_sum = 7 + 3
after_multiplication = parenthesized_sum * 10
mixed_result = after_multiplication / 5
print('(7 + 3) * 10 / 5 -> 10 * 10 / 5 -> 100 / 5 ->', mixed_result)
```

```python
# Reasoning trace: Python can decide each result from the first operand.
print('False and ... -> False; the second operand is skipped:', False and (10 / 0))
print('True or ...   -> True;  the second operand is skipped:', True or (10 / 0))

# A practical guard: division runs only when denominator != 0 is True.
numerator, denominator = 10, 0
safe_test = (denominator != 0) and (numerator / denominator > 2)
print('Is division valid and quotient > 2?', safe_test)

# Ordinary Boolean expressions
age = 21
has_id = True
print((age >= 18) and has_id)
print((age < 18) or (not has_id))
```
